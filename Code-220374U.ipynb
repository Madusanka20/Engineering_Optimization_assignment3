```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({"font.size": 11, "axes.grid": True, "grid.alpha": 0.3})
cols = ["tab:blue", "tab:orange", "tab:green", "tab:red"]


# Q1(a): L(mu,x) = x^2 + mu(1-x) vs x for mu = 0,1,2,3

x = np.linspace(-1, 3.5, 400)
fig, ax = plt.subplots(figsize=(6.5, 4.2))
for mu, c in zip([0, 1, 2, 3], cols):
    L = x**2 + mu * (1 - x)
    ax.plot(x, L, c=c, label=fr"$\mu={mu}$")
    xm = mu / 2                                   # minimizer x(mu) = mu/2
    ax.plot(xm, xm**2 + mu * (1 - xm), "o", c=c, ms=8, mec="k")
    ax.annotate(fr"$x({mu})={xm:g}$", (xm, xm**2 + mu * (1 - xm)),
                textcoords="offset points", xytext=(0, -16),
                ha="center", fontsize=9)
ax.axvline(1, c="gray", ls="--", lw=1)
ax.text(1.03, 6.3, "$x=1$ (boundary)", fontsize=9, color="gray")
ax.set_xlabel("$x$"); ax.set_ylabel(r"$L(\mu,x)$")
ax.set_title("Q1(a)  $L(\\mu,x)$ for $\\mu=0,1,2,3$")
ax.set_ylim(-0.8, 7); ax.legend()
fig.tight_layout(); fig.savefig("fig1a.png", dpi=200)
plt.show();


# Q1(b): dual function d(mu) = mu - mu^2/4
mu = np.linspace(0, 4, 300)
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(mu, mu - mu**2 / 4, c="tab:blue", label=r"$d(\mu)=\mu-\mu^2/4$")
ax.plot(2, 1, "o", c="tab:red", ms=9)
ax.annotate(r"$(\mu^*,d^*)=(2,1)$", (2, 1),
            textcoords="offset points", xytext=(8, -18))
ax.axhline(1, c="gray", ls="--", lw=1, label=r"$p^*=1$")
ax.set_xlabel(r"$\mu$"); ax.set_ylabel(r"$d(\mu)$")
ax.set_title("Q1(b)  Dual function")
ax.legend(loc="lower center")
fig.tight_layout(); fig.savefig("fig1b.png", dpi=200)
plt.show();


# Q1(c) first plot: L(mu,x) vs mu for several fixed x

mu = np.linspace(0, 5, 200)
fig, ax = plt.subplots(figsize=(6.5, 4.2))
for xv, c in zip([-1, 0, 0.5, 1, 2, 3],
                 ["tab:red", "tab:orange", "gold", "k", "tab:green", "tab:blue"]):
    ax.plot(mu, xv**2 + mu * (1 - xv), c=c, label=f"$x={xv:g}$")
ax.set_ylim(-1, 12)
ax.set_xlabel(r"$\mu$"); ax.set_ylabel(r"$L(\mu,x)$")
ax.set_title("Q1(c)  $L(\\mu,x)$ versus $\\mu$ for fixed $x$")
ax.legend(ncol=2, fontsize=9)
fig.tight_layout(); fig.savefig("fig1c1.png", dpi=200)
plt.show(); 


# Q1(c) second plot: primal function g(x) on [-1, 3]

fig, ax = plt.subplots(figsize=(6.5, 4.2))
xf = np.linspace(1, 3, 100)
ax.plot(xf, xf**2, c="tab:blue", lw=2, label=r"$g(x)=x^2$ (feasible)")
ax.plot(1, 1, "o", c="tab:blue")
for xi in [0.5, 0, -0.5, -1]:                     # arrows = +infinity
    ax.annotate("", xy=(xi, 9), xytext=(xi, 6),
                arrowprops=dict(arrowstyle="->", color="tab:red", lw=2))
ax.plot([], [], c="tab:red", lw=2, label=r"$g(x)=+\infty$ (infeasible)")
ax.axvline(1, c="tab:red", ls="--", lw=1)
ax.set_xlim(-1.2, 3.2); ax.set_ylim(0, 10)
ax.set_xlabel("$x$"); ax.set_ylabel("$g(x)$")
ax.set_title("Q1(c)  Primal function $g(x)$")
ax.legend(loc="lower right")
fig.tight_layout(); fig.savefig("fig1c2.png", dpi=200)
plt.show(); 


# Q2(a)(iv): L(mu,x*) vs mu  and  L(mu*,x) vs x

fig, (a1, a2) = plt.subplots(1, 2, figsize=(8, 3.4))

m = np.linspace(0, 4, 100)
a1.plot(m, np.ones_like(m), c="tab:blue")
a1.plot(2, 1, "o", c="tab:red", ms=8)
a1.annotate(r"$L(\mu^*,x^*)=1$", (2, 1),
            textcoords="offset points", xytext=(-10, 12), fontsize=9)
a1.set_ylim(0, 2)
a1.set_xlabel(r"$\mu$"); a1.set_ylabel(r"$L(\mu,x^*)$")
a1.set_title(r"$L(\mu,x^*)=1$")

xx = np.linspace(-1, 3, 200)
a2.plot(xx, (xx - 1)**2 + 1, c="tab:blue")
a2.plot(1, 1, "o", c="tab:red", ms=8)
a2.annotate(r"$L(\mu^*,x^*)=1$", (1, 1),
            textcoords="offset points", xytext=(8, -18), fontsize=9)
a2.set_xlabel("$x$"); a2.set_ylabel(r"$L(\mu^*,x)$")
a2.set_title(r"$L(\mu^*,x)=(x-1)^2+1$")

fig.tight_layout(); fig.savefig("fig2a.png", dpi=200)
plt.show(); 


# Q3(b): lines L(mu,x) for x = 0,1,2, lower envelope, true d(mu)

mu = np.linspace(0, 4, 300)
fig, ax = plt.subplots(figsize=(6.5, 4.2))
labels = {0: r"$x=0$: $L=\mu$", 1: r"$x=1$: $L=1$", 2: r"$x=2$: $L=4-\mu$"}
for xv, c in zip([0, 1, 2], ["tab:green", "tab:orange", "tab:purple"]):
    ax.plot(mu, xv**2 + mu * (1 - xv), c=c, lw=1.3, label=labels[xv])
env = np.minimum(np.minimum(mu, 1), 4 - mu)       # lower envelope
ax.plot(mu, env, c="k", lw=3, label="lower envelope")
ax.plot(mu, mu - mu**2 / 4, "--", c="tab:red", label=r"true $d(\mu)$")
ax.set_ylim(-0.3, 4.2)
ax.set_xlabel(r"$\mu$"); ax.set_ylabel(r"$L(\mu,x)$")
ax.set_title("Q3(b)  Lower envelope of $L(\\mu,x)$")
ax.legend(fontsize=9)
fig.tight_layout(); fig.savefig("fig3b.png", dpi=200)
plt.show(); 
```
